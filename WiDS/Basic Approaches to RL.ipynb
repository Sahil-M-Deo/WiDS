Give pseudocode for a complete algorithm for the n-armed
bandit problem. Use greedy action selection and incremental computation of
action values with  = 1/k step-size parameter. Assume a function bandit(a)
that takes an action and returns a reward. Use arrays and variables; do not
subscript anything by the time index t (for examples of this style of pseudocode,
see Figures 4.1 and 4.3). Indicate how the action values are initialized
and updated after each reward. Indicate how the step-size parameters are set
for each action as a function of how many times it has been tried.

```python
import numpy as np
Method1="Greedy"
Method2="Epsilon Greedy"
Method3="UCB"
runs=100
tot_reward=0
tot_reward1=0
tot_reward2=0
tot_reward3=0

for __ in range(runs):
    t=100000
    n=10
    #t is the number of times we can play 
    #n is number of bandits

    mean=np.random.rand(n) #generating random bandits

    def bandit(a):
        return 2*mean[a]*np.random.rand()  
    
    #Our method1 implemented:
    def method1():
        averages=10*np.ones(n); #optimistic initialization for greedy approach
        no_of_times=np.zeros(n) #count of number of times each bandit is played
        total=0
        for _ in range(t):
            best_actions = np.flatnonzero(averages==averages.max())
            action = np.random.choice(best_actions)
            reward=bandit(action)
            total+=reward
            no_of_times[action]+=1
            averages[action]+=(reward-averages[action])/no_of_times[action]
        return total
    
    tot_reward1+=method1()

    #Our method2 implemented:
    def method2():
        total=0
        eps=0.1 #epsilon value 

        averages=np.zeros(n); #initialization
        no_of_times=np.zeros(n) #count of number of times each bandit is played
        for _ in range(t):
            if(np.random.rand()<eps):
                action=np.random.randint(0,n)
            else:
                best_actions = np.flatnonzero(averages==averages.max())
                action=np.random.choice(best_actions)
            reward=bandit(action)
            total+=reward
            no_of_times[action]+=1
            averages[action]+=(reward-averages[action])/no_of_times[action]
        return total
    
    tot_reward2+=method2()

    def method3():
        total=0
        c=2
        averages=np.zeros(n); #initialization
        no_of_times=np.zeros(n) #count of number of times each bandit is played
        for i in range(n):
            reward=bandit(i)
            averages[i]=reward
            no_of_times[i]=1
            total+=reward
    
        for _ in range(n, t):  # start from n since we already pulled each arm once
            ucb=averages+np.sqrt(c*np.log(t+1)/no_of_times)
            best_actions = np.flatnonzero(ucb==ucb.max())
            action=np.random.choice(best_actions)
            reward=bandit(action)
            total+=reward
            no_of_times[action]+=1
            averages[action]+=(reward-averages[action])/no_of_times[action]
        return total
    
    tot_reward3+=method3()

    #below is random selection method as a reference
    def method():
        total=0
        for _ in range(t):
            total+=bandit(np.random.randint(0, n))
        return total
    
    tot_reward+=method()
print(Method1, "performs better by ",100*tot_reward1/tot_reward, "percent")
print(Method2, "performs better by ",100*tot_reward2/tot_reward, "percent")
print(Method3, "performs better by ",100*tot_reward3/tot_reward, "percent")
```
